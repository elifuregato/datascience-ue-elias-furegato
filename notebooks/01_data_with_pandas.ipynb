# Data with Pandas

```python
import pandas as pd
from matplotlib import pyplot as plt
```

```python
from pathlib import Path
csv_path = Path("../data/eistage.csv")

df = pd.read_csv(
    csv_path,
    sep=";",                
    skiprows=[0, 1],             
    low_memory=False,       
    encoding="utf-8",       
    usecols=[3, 5, 6],      
    nrows=70,
    names= ["year", "heat_days", "ice_days"],
    parse_dates=["year"]           
)

print(df.shape)
print(df.dtypes)
df.head()
```
